###Ejemplo para ilustrar la correlacion en Analitica de datos

```python
# ==============================================
# EJEMPLO DE CORRELACIÓN Y MAPAS DE CORRELACIÓN
# ==============================================

# Importamos las librerías necesarias
import pandas as pd                  # Para manejar los datos en formato de tabla
import seaborn as sns                # Para visualizaciones estadísticas (heatmap)
import matplotlib.pyplot as plt      # Para gráficos en general
from sklearn.datasets import load_iris   # Dataset de ejemplo (flores iris)

# -----------------------------
# 1. Cargar un dataset
# -----------------------------
# Usamos el dataset Iris, que trae medidas de flores (variables aleatorias continuas)
iris = load_iris()

# Convertimos el dataset en un DataFrame de pandas para trabajar más fácil
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# -----------------------------
# 2. Calcular la matriz de correlación
# -----------------------------
# Usamos el método .corr() para calcular la correlación entre todas las variables
# Por defecto, calcula la correlación de Pearson (dependencia lineal)
correlation_matrix = df.corr()

# Mostramos la matriz de correlación numérica en consola
print("Matriz de correlación:\n", correlation_matrix)

# -----------------------------
# 3. Visualizar la matriz con un mapa de calor (heatmap)
# -----------------------------
plt.figure(figsize=(8,6))  # Definimos el tamaño del gráfico

# Creamos un mapa de calor usando seaborn
sns.heatmap(correlation_matrix,
            annot=True,        # Muestra los valores numéricos dentro de las celdas
            cmap="coolwarm",   # Colores: azul (negativo) → rojo (positivo)
            fmt=".2f")         # Formato de los números con 2 decimales

plt.title("Mapa de Correlación (Iris Dataset)")  # Título del gráfico
plt.show()

# -----------------------------
# 4. Ejemplo de correlación con diagrama de dispersión
# -----------------------------
# Escogemos dos variables aleatorias: largo del sépalo y largo del pétalo
plt.figure(figsize=(6,4))
sns.scatterplot(x=df["sepal length (cm)"],
                y=df["petal length (cm)"])

plt.title("Diagrama de dispersión: Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.show()

# Con este scatterplot podemos ver gráficamente la correlación positiva entre estas dos variables
```
