## **Módulo 1: Clases, Listas y Estructuras de Datos**

1. Crea una lista vacía y agrega cinco números enteros utilizando `append()`.
2. Convierte la lista anterior en una tupla y explica la diferencia entre ambas estructuras.
3. Implementa una clase llamada `Estudiante` con los atributos: `nombre`, `edad` y `carrera`.
   - Agrega un método que permita mostrar los datos en formato: `Nombre - Edad años - Carrera`.
   - Crea al menos 3 objetos de esta clase y guárdalos en una lista.
4. Escribe una función que reciba una lista de estudiantes y retorne el promedio de edades.

````python
# Módulo 1: Clases, Listas y Estructuras de Datos

# Punto 1: Crear una lista vacía y agregar cinco números enteros
# Inicializamos una lista vacía
mi_lista = []

# Agregamos cinco números enteros a la lista usando append()
mi_lista.append(1)
mi_lista.append(2)
mi_lista.append(3)
mi_lista.append(4)
mi_lista.append(5)

# Imprimimos la lista resultante
print("Lista después de agregar elementos:", mi_lista)

# Punto 2: Convertir la lista en una tupla
# Convertimos la lista 'mi_lista' a una tupla usando la función tuple()
mi_tupla = tuple(mi_lista)

# Imprimimos la tupla resultante
print("Tupla convertida desde la lista:", mi_tupla)

# Explicación de la diferencia entre listas y tuplas (se incluirá en celdas Markdown separadas)
````

Una lista y una tupla son estructuras de datos que me permiten guardar diferentes elementos, pero las a las listas se les pueden modificar el contenido (mutables), pero a las tuplas no (inmutables). Ademas, su sintaxis es diferente: las listas se definen entre corchetes y las tuplas entre parentesis.

````python
#Punto 3: Implementar una clase llamada Estudiante

#Definimos la clase Estudiante
class Estudiante:
    # El método __init__ es el constructor de la clase
    # Se llama automáticamente al crear un nuevo objeto de la clase
    def __init__(self, nombre, edad, carrera):
        # Inicializamos los atributos del objeto con los valores pasados al constructor
        self.nombre = nombre  # Atributo para el nombre del estudiante
        self.edad = edad      # Atributo para la edad del estudiante
        self.carrera = carrera # Atributo para la carrera del estudiante

    # Método para mostrar los datos del estudiante en un formato específico
    def mostrar_datos(self):
        # Retorna una cadena formateada con la información del estudiante
        return f"{self.nombre} - {self.edad} años - {self.carrera}"

# Creamos al menos 3 objetos de la clase Estudiante
estudiante_1 = Estudiante("Ana", 18, "Ingeniería de Sistemas")
estudiante_2 = Estudiante("Juan", 21, "Administración de Empresas")
estudiante_3 = Estudiante("Pedro", 22, "Derecho")

# Guardamos los objetos de la clase Estudiante en una lista
lista_estudiantes = [estudiante_1, estudiante_2, estudiante_3]

# Imprimimos los datos de cada estudiante en la lista usando el método mostrar_datos()
print("Datos de los estudiantes:\n")
for estudiante in lista_estudiantes:
    print(estudiante.mostrar_datos())
````

````python
#Punto 4: Función que reciba una lista de estudiantes y retorne el promedio de edades.

def calcular_promedio_edades(lista_estudiantes):
    # Inicializamos la suma de las edades en cero
    suma_edades = 0
    # Iteramos a través de la lista de estudiantes
    for estudiante in lista_estudiantes:
        # Sumamos la edad de cada estudiante a la suma total
        suma_edades += estudiante.edad

    # Verificamos si la lista de estudiantes no está vacía para evitar división por cero
    if len(lista_estudiantes) > 0:
        # Calculamos el promedio dividiendo la suma total por el número de estudiantes
        promedio = suma_edades / len(lista_estudiantes)
        return promedio
    else:
        # Si la lista está vacía, retornamos 0 como promedio
        return 0

# Ejemplo de uso de la función con la lista de estudiantes creada anteriormente
promedio_edades = calcular_promedio_edades(lista_estudiantes)
print(f"\nEl promedio de edad de los estudiantes es: {promedio_edades:.2f} años")
````

## **Módulo 2: Manejo de Datos con Pandas**

1. Crea un **DataFrame** con la siguiente información:

| Nombre  | Edad | Ciudad   |
|---------|------|----------|
| Ana     | 23   | Cali     |
| Juan    | 21   | Medellín |
| Pedro   | 25   | Bogotá   |
| Laura   | 22   | Cali     |

2. Obtén:
- El número de filas y columnas del DataFrame.
- La lista de ciudades únicas.
- La edad promedio de los estudiantes.

3. Agrega una nueva columna llamada `Beca` que sea `True` si la edad es menor a 23, y `False` en caso contrario.
4. Exporta el DataFrame a un archivo `.csv`.

````python
import pandas as pd

# Punto 1: Crear un DataFrame con la información proporcionada

# Datos para el DataFrame en un diccionario
datos = {
    'Nombre': ['Ana', 'Juan', 'Pedro', 'Laura'],
    'Edad': [23, 21, 25, 22],
    'Ciudad': ['Cali', 'Medellín', 'Bogotá', 'Cali']
}

# Crear el DataFrame
df_estudiantes = pd.DataFrame(datos)

# Mostrar el DataFrame creado
print("DataFrame de Estudiantes:\n")
print(df_estudiantes)
````

````python
#PUNTO 2: obtener el número de filas y columnas, la lista de ciudades únicas y la edad promedio de los estudiantes

# Obtener el número de filas y columnas
num_filas, num_columnas = df_estudiantes.shape
print(f"\nNúmero de filas: {num_filas}")
print(f"Número de columnas: {num_columnas}")

# Obtener la lista de ciudades únicas
ciudades_unicas = df_estudiantes['Ciudad'].unique()
print(f"\nCiudades únicas: {list(ciudades_unicas)}")

# Obtener la edad promedio de los estudiantes
edad_promedio = df_estudiantes['Edad'].mean()
print(f"\nEdad promedio de los estudiantes: {edad_promedio:.2f} años")
````

````python
#PUNTO 3: nueva columna llamada Beca basada en la edad

# Agregamos la columna 'Beca'
df_estudiantes['Beca'] = df_estudiantes['Edad'] < 23

# Mostramos el DataFrame con la nueva columna
print("\nDataFrame con la columna 'Beca':")
print(df_estudiantes)
````

````python
#PUNTO 4: Exportar el DataFrame a un archivo .csv

# Exportamos el DataFrame a un archivo CSV
# index=False evita escribir el índice del DataFrame en el archivo CSV
df_estudiantes.to_csv('estudiantes.csv', index=False)

print("\nDataFrame exportado exitosamente a 'estudiantes.csv'")

# Descargar el archivo localmente
from google.colab import files
#Quitar el # si se quiere descargar el archivo CSV.
#files.download('estudiantes.csv')
````

## **Módulo 3: Práctica de Matrices y Ejercicios Numéricos**

1. Crea una matriz **diagonal** de 5x5 a partir del vector `[1,2,3,4,5]` sin usar funciones predefinidas de Numpy.
2. Implementa una función que reciba dos matrices y retorne su producto.
3. Genera una matriz de 5x5 con valores aleatorios entre 0 y 9 e identifica el valor máximo y mínimo junto con sus posiciones.
4. Usando `numpy`, escribe un código que calcule la **transpuesta** y la **inversa** de una matriz cuadrada aleatoria de 3x3.

````python
# PUNTO 1: Crear una matriz diagonal de 5x5 a partir del vector [1,2,3,4,5] sin usar funciones predefinidas de Numpy.

# Definimos el vector del cual obtendremos los elementos de la diagonal.
vector = [1, 2, 3, 4, 5]

# Obtenemos la longitud del vector para determinar el tamaño de la matriz cuadrada (n x n).
n = len(vector)

# Inicializamos una lista vacía que representará nuestra matriz.
matriz_diagonal = []

# Iteramos sobre las filas de la matriz (desde 0 hasta n-1).
for i in range(n):
    fila = [] # Inicializamos una lista vacía para representar la fila actual.
    # Iteramos sobre las columnas de la matriz (desde 0 hasta n-1).
    for j in range(n):
        # Verificamos si estamos en la diagonal principal (donde el índice de fila 'i' es igual al índice de columna 'j').
        if i == j:
            # Si estamos en la diagonal, agregamos el elemento correspondiente del vector.
            fila.append(vector[i])
        else:
            # Si no estamos en la diagonal, agregamos un cero a la fila.
            fila.append(0)
    # Una vez que hemos construido la fila completa, la agregamos a la lista que representa la matriz.
    matriz_diagonal.append(fila)

# Imprimir la matriz diagonal resultante.
print("Matriz diagonal:")
for fila in matriz_diagonal:
    print(fila)
````

````python
#PUNTO 2: Función que reciba dos matrices y retorne su producto.

def multiplicar_matrices(matriz1, matriz2):
    # Obtenemos las dimensiones de las matrices
    filas_matriz1 = len(matriz1)
    columnas_matriz1 = len(matriz1[0])
    filas_matriz2 = len(matriz2)
    columnas_matriz2 = len(matriz2[0])

    # Verificamos si las matrices se pueden multiplicar
    # El número de columnas de la primera matriz debe ser igual al número de filas de la segunda matriz
    if columnas_matriz1 != filas_matriz2:
        return "Error: Las matrices no se pueden multiplicar. El número de columnas de la primera matriz debe ser igual al número de filas de la segunda matriz."

    # Creamos la matriz resultado inicializada con ceros
    # La matriz resultado tendrá el número de filas de la primera matriz y el número de columnas de la segunda matriz
    matriz_resultado = [[0 for _ in range(columnas_matriz2)] for _ in range(filas_matriz1)]

    # Realizamos la multiplicación de matrices
    # Iteramos sobre las filas de la primera matriz
    for i in range(filas_matriz1):
        # Iteramos sobre las columnas de la segunda matriz
        for j in range(columnas_matriz2):
            # Iteramos sobre los elementos de la fila de la primera matriz y la columna de la segunda matriz
            for k in range(columnas_matriz1): # o filas_matriz2, ya que son iguales
                matriz_resultado[i][j] += matriz1[i][k] * matriz2[k][j]

    return matriz_resultado

# Ejemplo de uso de la función
matriz_a = [[1, 2, 3],
            [4, 5, 6]]

matriz_b = [[7, 8],
            [9, 10],
            [11, 12]]

producto = multiplicar_matrices(matriz_a, matriz_b)

print("Matriz A:")
for fila in matriz_a:
    print(fila)

print("\nMatriz B:")
for fila in matriz_b:
    print(fila)

print("\nProducto de las matrices A y B:")
if isinstance(producto, str):
    print(producto)
else:
    for fila in producto:
        print(fila)

# Ejemplo de matrices que no se pueden multiplicar
matriz_c = [[1, 2],
            [3, 4]]

matriz_d = [[5, 6],
            [7, 8],
            [9, 10]]

producto_error = multiplicar_matrices(matriz_c, matriz_d)
print("\nProducto de las matrices C y D (ejemplo de error):")
print(producto_error)
````

````python
#PUNTO 3: matriz de 5x5 con valores aleatorios entre 0 y 9 e identifica el valor máximo y mínimo junto con sus posiciones.

import random

# Generar una matriz de 5x5 con valores aleatorios entre 0 y 9
matriz_aleatoria = []
for i in range(5):
    fila = []
    for j in range(5):
        # Genera un entero aleatorio entre 0 y 9 (inclusive)
        fila.append(random.randint(0, 9))
    matriz_aleatoria.append(fila)

print("Matriz Aleatoria de 5x5:")
for fila in matriz_aleatoria:
    print(fila)

# Inicializar variables para el valor máximo, mínimo y sus posiciones
valor_maximo = -1 # Inicializamos con un valor menor que cualquier posible valor en la matriz
posicion_maximo = ( -1, -1) # Inicializamos con posiciones inválidas

valor_minimo = 10 # Inicializamos con un valor mayor que cualquier posible valor en la matriz
posicion_minimo = (-1, -1) # Inicializamos con posiciones inválidas

# Recorrer la matriz para encontrar el valor máximo y mínimo y sus posiciones
for i in range(5):
    for j in range(5):
        valor_actual = matriz_aleatoria[i][j]

        # Comprobar si el valor actual es mayor que el máximo encontrado hasta ahora
        if valor_actual > valor_maximo:
            valor_maximo = valor_actual
            posicion_maximo = (i, j)

        # Comprobar si el valor actual es menor que el mínimo encontrado hasta ahora
        if valor_actual < valor_minimo:
            valor_minimo = valor_actual
            posicion_minimo = (i, j)

# Imprimir los resultados
print(f"\nValor máximo encontrado: {valor_maximo}")
print(f"Posición del valor máximo: {posicion_maximo}")
print(f"\nValor mínimo encontrado: {valor_minimo}")
print(f"Posición del valor mínimo: {posicion_minimo}")
````

````python
#PUNTO 4:  código que calcule la transpuesta y la inversa de una matriz cuadrada aleatoria de 3x3 usando Numpy.

import numpy as np

# Generar una matriz cuadrada aleatoria de 3x3
# np.random.rand(3, 3) crea una matriz de 3x3 con valores aleatorios entre 0 y 1
# Multiplicamos por 10 y convertimos a entero para tener valores entre 0 y 9
matriz_original = np.random.randint(0, 10, size=(3, 3))


print("Matriz Original de 3x3:")
print(matriz_original)

# Calcular la transpuesta de la matriz
# La transpuesta se obtiene intercambiando filas por columnas
matriz_transpuesta = np.transpose(matriz_original)
print("\nMatriz Transpuesta:")
print(matriz_transpuesta)

# Calcular la inversa de la matriz
# La inversa de una matriz A es una matriz B tal que A * B = I (matriz identidad)
# Solo las matrices cuadradas no singulares (determinante distinto de cero) tienen inversa.
try:
    matriz_inversa = np.linalg.inv(matriz_original)
    print("\nMatriz Inversa:")
    print(matriz_inversa)
except np.linalg.LinAlgError:
    # Capturamos el error si la matriz es singular y no tiene inversa
    print("\nLa matriz es singular y no tiene inversa.")
````

````python
#PUNTO 1: Grafica la función y = x^2 para x en el rango de -10 a 10.
import matplotlib.pyplot as plt

#Importamos numpy para generar el rango de valores de x
import numpy as np

# Generar valores de x en el rango de -10 a 10
# np.linspace(inicio, fin, numero_de_puntos) crea un array de números espaciados uniformemente
x = np.linspace(-10, 10, 100) # Creamos 100 puntos entre -10 y 10 para una gráfica suave

# Calcular los valores correspondientes de y = x^2
y = x**2

# Crear la gráfica
plt.figure(figsize=(8, 6)) # Define el tamaño de la figura (opcional)
plt.plot(x, y, label='y = x^2', color='red') # Crea el gráfico de línea con x, y, etiqueta y color

# Añadir título y etiquetas a los ejes
plt.title('Gráfica de la función y = x^2')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')

# Añadir una cuadrícula (opcional)
plt.grid(True)

# Añadir una leyenda (muestra la etiqueta de la línea)
plt.legend()

# Mostrar la gráfica
plt.show()
````

````python
#PUNTO 2: gráfico de barras con los datos de la tabla

import matplotlib.pyplot as plt

# Datos de las ciudades y su población
Ciudades = ["Cali", "Bogotá", "Medellín", "Barranquilla"]
Poblacion = [2.2, 7.4, 2.5, 1.2]  # en millones

# Definir una lista de colores para cada barra
colores = ['blue', 'green', 'red', 'purple']

# Crear la gráfica de barras con colores diferentes
plt.figure()
plt.bar(Ciudades, Poblacion, color=colores) # Pasamos la lista de colores al argumento 'color'

# Añadir título y etiquetas a los ejes
plt.title('Población de las Ciudades')
plt.xlabel('Ciudades')
plt.ylabel('Población (millones)')

# Mostrar la gráfica
plt.show()
````

````python
#PUNTO 3:  gráfico de dispersión de 100 puntos aleatorios en el plano XY.

import matplotlib.pyplot as plt
import numpy as np

# Generar 100 puntos aleatorios en el plano XY
x = np.random.rand(100)
y = np.random.rand(100)

# Crear la gráfica de dispersión
plt.figure
plt.scatter(x, y, label='Puntos Aleatorios', color='blue', marker='o')

# Añadir título y etiquetas a los ejes
plt.title('Gráfico de Dispersión de 100 Puntos Aleatorios')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')

# Añadir una leyenda (opcional)
plt.legend()

# Mostrar la gráfica
plt.show()
````

````python
#PUNTO 4: Simulación de una serie temporal de 30 días con valores aleatorios y representación en una gráfica de líneas.

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd # Importamos pandas para manejar fechas más fácilmente

# Generar una serie temporal de 30 días
# Creamos un rango de fechas de 30 días a partir de hoy
fechas = pd.date_range(start=pd.to_datetime('today').normalize(), periods=30, freq='D')

# Generar valores aleatorios para cada día (simulando una serie temporal)
# np.random.rand(30) crea un array de 30 valores aleatorios entre 0 y 1
valores = np.random.rand(30) * 100 # Multiplicamos por 100 para tener un rango de valores más amplio (ej. 0 a 100)

# Crear la gráfica de líneas para la serie temporal
plt.figure(figsize=(10, 6)) # Definimos un tamaño de figura adecuado para una serie temporal
plt.plot(fechas, valores, marker='o', linestyle='-', color='green', label='Serie Temporal Aleatoria') # Creamos el gráfico de líneas

# Añadir título y etiquetas a los ejes
plt.title('Simulación de Serie Temporal de 30 Días')
plt.xlabel('Fecha')
plt.ylabel('Valor')

# Formatear el eje x para mostrar las fechas de manera legible (opcional)
plt.gcf().autofmt_xdate() # Auto-formatea las etiquetas del eje x para que no se superpongan

# Añadir una cuadrícula (opcional)
plt.grid(True)

# Añadir una leyenda
plt.legend()

# Mostrar la gráfica
plt.show()
````
