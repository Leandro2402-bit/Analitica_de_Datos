```python
# ==============================
# Descriptores de datos en Python
# Ejemplo: [2, 4, 6, 8, 10, 12, 14, 16]
# ==============================

import numpy as np   # Importamos NumPy para cálculos numéricos

# Definimos los datos
datos = np.array([2, 4, 6, 8, 10, 12, 14, 16])

# ---- Media ----
media = np.mean(datos)   # Calculamos el promedio

# ---- Mediana ----
mediana = np.median(datos)   # Calculamos la mediana

# ---- Varianza ----
varianza = np.var(datos)   # Calculamos la varianza

# ---- Desviación estándar ----
desviacion = np.std(datos)  # Calculamos la desviación estándar
desviacion = round(desviacion, 2)  # Redondeamos a 2 decimales

# ---- Mostramos resultados ----
print("Datos:", datos)
print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Varianza: {varianza}")
print(f"Desviación estándar: {desviacion}")
```

```python
# ==============================
# Cuartiles, Rango Intercuartílico y Boxplot en Python
# Ejemplo: [2, 4, 6, 8, 10, 12, 14, 16]
# ==============================

import numpy as np
import matplotlib.pyplot as plt   # Librería para graficar

# Definimos los datos
datos = np.array([2, 4, 6, 8, 10, 12, 14, 16])

# ---- Cuartiles ----
Q1 = np.percentile(datos, 25)   # Primer cuartil (25%)
Q2 = np.percentile(datos, 50)   # Segundo cuartil (mediana, 50%)
Q3 = np.percentile(datos, 75)   # Tercer cuartil (75%)

# ---- Rango intercuartílico (IQR) ----
IQR = Q3 - Q1

# ---- Mostramos resultados ----
print("Datos:", datos)
print(f"Q1 (25%): {Q1}")
print(f"Q2 (Mediana, 50%): {Q2}")
print(f"Q3 (75%): {Q3}")
print(f"Rango Intercuartílico (IQR): {IQR}")

# ---- Graficamos el Boxplot ----
plt.boxplot(datos, vert=False, patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Boxplot de los datos")
plt.xlabel("Valores")
plt.show()
```

```python
# ==============================
# Normalización de datos en Python
# Métodos: Min-Max y Z-score
# Ejemplo: [2, 4, 6, 8, 10, 12, 14, 16]
# ==============================

import numpy as np
import pandas as pd   # Para mostrar resultados en tabla

# Definimos los datos
datos = np.array([2, 4, 6, 8, 10, 12, 14, 16])

# ---- Min-Max Normalization ----
x_min = np.min(datos)   # Valor mínimo
x_max = np.max(datos)   # Valor máximo
min_max_norm = (datos - x_min) / (x_max - x_min)   # Fórmula

# ---- Z-score Normalization ----
media = np.mean(datos)          # Media de los datos
desviacion = np.std(datos)      # Desviación estándar
z_score_norm = (datos - media) / desviacion   # Fórmula

# ---- Mostramos resultados en tabla ----
tabla = pd.DataFrame({
    "Original": datos,
    "Min-Max": np.round(min_max_norm, 2),   # Redondeamos a 2 decimales
    "Z-score": np.round(z_score_norm, 2)    # Redondeamos a 2 decimales
})

print(tabla)
```
